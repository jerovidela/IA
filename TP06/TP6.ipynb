## Ejercicios Teóricos

1. ¿En qué tipo de algoritmos se basa un planificador para encontrar el mejor camino a un estado solución?

Un planificador se basa en algoritmos de búsqueda, aplicados sobre un modelo lógico del problema. Utiliza representaciones de estados, acciones y objetivos en forma de literales, y busca una secuencia de acciones que transforme el estado inicial en uno que cumpla el objetivo. Las acciones tienen precondiciones y efectos, y todo el proceso se formaliza con herramientas como STRIPS. Este enfoque permite al agente construir planes complejos de forma eficiente, sobre todo en entornos deterministas y totalmente observables.


2. ¿Qué tres elementos se encuentran dentro de una acción formulada en lenguaje STRIPS? Describa brevemente qué función cumple cada uno.

Una acción en lenguaje STRIPS tiene tres elementos principales:

- Nombre y parámetros: Identifica la acción y las variables que intervienen.
Ejemplo: Volar(p, desde, hasta) indica que se trata de una acción de vuelo con un avión p desde un lugar a otro.

- Precondiciones: Son las condiciones que deben cumplirse en el estado actual para que la acción pueda ejecutarse.
Ejemplo: En(p, desde) ∧ Avión(p) indica que el avión debe estar en el lugar de origen.

- Efectos: Describen cómo cambia el estado del mundo después de ejecutar la acción.
Ejemplo: ¬En(p, desde) ∧ En(p, hasta) indica que el avión ya no está en el lugar de origen y ahora está en el destino.

3. Describa las ventajas y desventajas de desarrollar un algoritmo de planificación hacia adelante y hacia atrás en el espacio de estados.


1. Planificación hacia adelante (forward search):
Parte del estado inicial y aplica acciones disponibles hasta llegar a un estado que cumpla el objetivo.

- Ventajas:

    Es directa: el estado inicial está completamente definido.

    Es fácil verificar qué acciones son aplicables.

    Útil cuando se conocen bien las condiciones iniciales y se puede simular el entorno.

- Desventajas:

    Puede generar muchos estados irrelevantes para el objetivo.

    Difícil mantener el enfoque en lo que realmente se quiere lograr.

2. Planificación hacia atrás (backward search):
Parte del objetivo y busca acciones que puedan lograrlo, retrocediendo hasta llegar al estado inicial.

- Ventajas:

    Se enfoca en las acciones necesarias para cumplir el objetivo, evitando caminos inútiles.

    Puede ser más eficiente cuando los objetivos son muy específicos.

- Desventajas:

    Es más complejo verificar si las precondiciones de las acciones se cumplen en algún estado posible.

    Difícil de aplicar si no se conoce bien el estado inicial o si el dominio tiene muchas acciones posibles.

6. Suponga que tiene un robot de oficina capaz de moverse y tomar y depositar objetos. El robot solo puede tener un objeto a la vez, pero puede conseguir una *caja* en la que depositar varios objetos. Suponga que programa al robot para *ir a la tienda* a comprarle un *café* y en el camino de vuelta tome una *carta* del *buzón* de la oficina para para que se la traiga junto con el café. Describa en lenguaje STRIPS:

6.1 El dominio del robot (nombre, predicados y acciones que puede hacer el robot).

6.2 El problema que se quiere resolver (estado inicial, estado objetivo y objetos del mundo representados).

6.3 Introduzca el código desarrollado en los puntos anteriores en el [planificador online](http://lcas.lincoln.ac.uk/fast-downward/) y obtenga el plan de acción que tomará el robot para cumplir lo solicitado.

```python
6.1-

(define (domain robot-strips)
  (:predicates
    (habitacion ?hab)
    (objeto ?obj)
    (robot-en ?hab)
    (en ?obj ?hab)
    (libre)
    (llevando ?obj)
    (en-caja ?obj)
    (conectado ?a - hab ?b - hab)
  )

  (:action moverse
    :parameters (?from ?to)
    :precondition (and (conectado ?from ?to) (robot-en ?from))
    :effect (and (robot-en ?to)
                 (not (robot-en ?from)))
  )

  (:action agarrar
    :parameters (?obj ?hab)
    :precondition (and (objeto ?obj) (habitacion ?hab)
                       (en ?obj ?hab) (robot-en ?hab) (libre))
    :effect (and (llevando ?obj)
                 (not (en ?obj ?hab))
                 (not (libre)))
  )

  (:action soltar
    :parameters (?obj ?hab)
    :precondition (and (objeto ?obj) (habitacion ?hab)
                       (llevando ?obj) (robot-en ?hab))
    :effect (and (en ?obj ?hab)
                 (libre)
                 (not (llevando ?obj)))
  )

  (:action meter_en_caja
    :parameters (?obj ?hab)
    :precondition (and (objeto ?obj) (habitacion ?hab)
                       (llevando ?obj) (robot-en ?hab)
                       (en caja ?hab))
    :effect (and (en-caja ?obj)
                 (libre)
                 (not (llevando ?obj)))
  )

  (:action sacar_de_caja
    :parameters (?obj ?hab)
    :precondition (and (objeto ?obj) (habitacion ?hab)
                       (en caja ?hab) (en-caja ?obj) (robot-en ?hab)
                       (libre))
    :effect (and (llevando ?obj)
                 (not (en-caja ?obj))
                 (not (libre)))
  )
)

6.2- 

(define (problem strips-robot-x-1)
  (:domain robot-strips)
  (:objects
    oficina tienda buzon
    cafe carta caja; 
  )
  (:init
    (habitacion oficina)
    (habitacion tienda)
    (habitacion buzon)
    (objeto cafe)
    (objeto carta)
    (objeto caja)
    (robot-en oficina)
    (libre)
    (en cafe tienda)
    (en carta buzon)
    (en caja oficina)
    (conectado oficina tienda)
    (conectado tienda buzon)
    (conectado buzon oficina)
  )
  (:goal
    (and (en caja oficina)
         (en-caja cafe)
         (en-caja carta))
  )
)

6.3 - 

(moverse oficina tienda)
(moverse tienda buzon)
(agarrar carta buzon)
(moverse buzon oficina)
(meter_en_caja carta oficina)
(moverse oficina tienda)
(agarrar cafe tienda)
(moverse tienda buzon)
(moverse buzon oficina)
(meter_en_caja cafe oficina)
; cost = 10 (unit cost)
```
