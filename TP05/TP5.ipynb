## Ejercicios Teóricos

1. ¿Cuáles son los tres axiomas de Kolmogorov?

- - Los tres axiomas de Kolmogórov son: la no negatividad, que establece que la probabilidad de cualquier evento es mayor o igual a cero; la normalización, que indica que la probabilidad del espacio muestral completo es igual a uno; y la aditividad, que señala que si dos o más eventos son mutuamente excluyentes, la probabilidad de su unión es igual a la suma de sus probabilidades individuales.


2. Una fábrica de clavos dispone de 2 máquinas que elaboran el 30% y 70% de los clavos que producen respectivamente. El porcentaje de clavos defectuosos de cada máquina es del 2% y 3%, respectivamente. Si se selecciona al azar un clavo de la producción y este fue defectuoso, ¿cuál es la probabilidad de que haya sido fabricado por la máquina?

Para poder calcular la probabilidad de que la máquina i haya fabricado el clavo defectuoso hay que utilizar teorema de Bayes.
P(Mi​/D)=(P(D/Mi​)*P(Mi​)) / P(D)

Para poder calcular esto falta el dato de la probabilidad de que el clavo este defectuoso, lo que calculamos:

P(D)=P(D/M1​)*P(M1​)+P(D∣M2​)*P(M2​) = 0.02*0.3+0.03*0.7 = 0.027

Con esta información ya se puede calcular la probabilidad de que la máquina i sea la que fabricó el clavo. Para poder reemplazar los valores vamos a calcular la probabilidad de que la máquina 1 haya fabricado el clavo

P(M1​/D)=(P(D/M1​)*P(M1​)) / P(D) = (0.02*0.3)/0.027 = 2/9 = 22.22%

3. La probabilidad de que un motor que sale de una fábrica con una *avería eléctrica* es de $10^{-3}$, y la probabilidad de que salga con una *avería mecánica* es de $10^{-5}$. Si existe un tipo de avería no se producen averías del otro tipo.
<br><br>Si el motor presenta *temperatura elevada* se enciende un *piloto luminoso* el 95% de las veces, cuando la *temperatura es reducida* el *piloto luminoso* se enciende el 99% de las veces, y a veces cuando la *temperatura se encuentra en un rango normal* el *piloto luminoso* se enciende erróneamente en un caso por millón.
<br><br>Cuando *no hay averías*, la *temperatura se eleva* en el 17% de los casos y es *reducida* el 5% de las veces. Si hay una *avería eléctrica*, la *temperatura se eleva* en el 90% de los casos y es *reducida* en el 1% de los casos. Finalmente cuando la *avería es mecánica*, la *temperatura está elevada* el 10% de los casos y *reducida* el 40% de las veces.
<br><br>Construya una Red Bayesiana y utilice inferencia por enumeración para calcular:

        3.1 La probabilidad de que el motor tenga una avería mecánica si se enciende el piloto.

Para calcular la probabilidad de que el motor presente una avería mecánica dado que el piloto está encendido, se aplicó el método de inferencia por enumeración.

Primero se construyó el árbol de probabilidades, partiendo del nodo raíz “Falla (F)”, que puede tomar los valores avería eléctrica (AE) con probabilidad 0.001, avería mecánica (AM) con 0.00001, y ninguna avería (AN) con 0.99899.

De cada una de estas ramas se desplegaron las posibles temperaturas (T). Por ejemplo, para el caso AM, se abren tres ramas: temperatura elevada (TE) con 0.10, temperatura reducida (TR) con 0.40, y temperatura normal (TN) con 0.50.

Finalmente, de cada valor de la temperatura se desprenden los estados del piloto (P): si la temperatura es elevada, el piloto se enciende con 0.95 y permanece apagado con 0.05; si la temperatura es reducida, se enciende con 0.99 y se apaga con 0.01; y si la temperatura es normal, se enciende solo con 0.000001 y se apaga con 0.999999.

Cada hoja del árbol representa el producto de las probabilidades acumuladas a lo largo de la rama. Por ejemplo:𝑃(𝐴𝑀,𝑇𝐸,𝑃𝐿)=𝑃(𝐴𝑀)⋅𝑃(𝑇𝐸∣𝐴𝑀)⋅𝑃(𝑃𝐿∣𝑇𝐸)=0.00001×0.10×0.95=9.5×10−7.

De forma análoga:

𝑃(𝐴𝑀,𝑇𝑅,𝑃𝐿)=0.00001×0.40×0.99=3.96×10−6,

𝑃(𝐴𝑀,𝑇𝑁,𝑃𝐿)=0.00001×0.50×0.000001=5.0×10−12.

La suma de estas ramas da:𝑃(𝐴𝑀,𝑃𝐿)=4.91×10−6.

El mismo procedimiento se aplica para las ramas correspondientes a AE y AN, obteniéndose:
𝑃(𝐴𝐸,𝑃𝐿)≈8.65×10−4,𝑃(𝐴𝑁,𝑃𝐿)≈0.210778.

En el paso final, se realiza la normalización mediante el factor 𝛼, definido como:𝛼=1/𝑃(𝐴𝑀,𝑃𝐿)+𝑃(𝐴𝐸,𝑃𝐿)+𝑃(𝐴𝑁,𝑃𝐿).

La probabilidad buscada resulta entonces: 𝑃(𝐴𝑀∣𝑃𝐿)=𝛼⋅𝑃(𝐴𝑀,𝑃𝐿)=4.91×10−6/0.211648≈2.3×10−5.

En otras palabras, la probabilidad de que el motor tenga una avería mecánica dado que el piloto se enciende es del orden de 0,0023%.

        3.2 La probabilidad de que el motor tenga una avería mecánica si se enciende el piloto y la temperatura es elevada.

En este segundo caso se buscó determinar la probabilidad de que el motor presente una avería mecánica dado que el piloto está encendido y la temperatura es elevada. Para resolverlo, se empleó nuevamente el método de inferencia por enumeración, adaptado a la evidencia observada.

A diferencia del caso anterior, en este escenario la información de la temperatura elevada (TE) ya restringe de manera significativa el árbol de inferencia. Como el nodo Piloto (P) depende únicamente de la temperatura, conocer que el piloto se encendió no agrega información adicional una vez fijada la temperatura. En consecuencia, el problema se reduce a calcular directamente la probabilidad de falla mecánica condicionada a la temperatura elevada.

El proceso comienza con el nodo raíz Falla (F), el cual contempla tres estados mutuamente excluyentes: avería mecánica (AM) con probabilidad 0.00001, avería eléctrica (AE) con 0.001 y ninguna avería (AN) con 0.99899. A partir de cada una de estas ramas se desprende la probabilidad de que la temperatura sea elevada. Para el caso de avería mecánica, la temperatura es elevada con probabilidad 0.10; para avería eléctrica, con 0.90; y en ausencia de avería, con 0.17.

Las probabilidades conjuntas se calculan multiplicando la probabilidad de la falla por la probabilidad de temperatura elevada en cada caso. Así se obtiene: 𝑃(𝐴𝑀,𝑇𝐸)=0.00001×0.10=1.0×10−6

𝑃(𝐴𝐸,𝑇𝐸)=0.001×0.90=9.0×10−4

𝑃(𝐴𝑁,𝑇𝐸)=0.99899×0.17≈0.169828

El siguiente paso consiste en calcular el denominador de la fracción bayesiana, que corresponde a la probabilidad total de observar una temperatura elevada, es decir, la suma de todas las ramas compatibles con dicha evidencia:

𝑃(𝑇𝐸)=𝑃(𝐴𝑀,𝑇𝐸)+𝑃(𝐴𝐸,𝑇𝐸)+𝑃(𝐴𝑁,𝑇𝐸)≈0.170729

Con este valor se determina el factor de normalización 𝛼=1/𝑃(𝑇𝐸), el cual garantiza que la distribución posterior sea válida.

Finalmente, se obtiene la probabilidad buscada al normalizar la rama de interés:

𝑃(𝐴𝑀∣𝑃𝐿,𝑇𝐸)=𝛼⋅𝑃(𝐴𝑀,𝑇𝐸)=1.0×10−6/0.170729≈5.86×10−6	

Este resultado indica que, aun en presencia de temperatura elevada y piloto encendido, la probabilidad de que la falla corresponda a una avería mecánica es extremadamente baja, del orden de 0,000586%.
